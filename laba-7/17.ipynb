```python
season_desc = pd.DataFrame({
    'season': [1, 2, 3, 4],
    'season_name': ['spring', 'summer', 'fall', 'winter']
})
```

```python
bike = bike.merge(season_desc, on='season', how='left')
bike
```

```python
bike[["season","count"]].groupby("season").count()
```

```python
p_t = bike.pivot_table("temp","humidity","windspeed")
p_t
```

```python
group = bike[["season","count"]].groupby("season").sum()
percentages = (group / group.sum()) * 100
percentages
```
